# 5: Data Preparation
We will create seven new attributes to facilitate data analysis and visualisation.

| New Attribute | Derived From | Purpose |
|---|---|---|
| `transaction_date` | `month` | Time-series analysis |
| `transaction_year` | `month` | Annual comparisons |
| `month_number` | `month` | Seasonal analysis |
| `remaining_lease_years` | `remaining_lease` | Lease-length analysis |
| `price_per_sqm` | `resale_price` / `floor_area_sqm` | More comparable resale-price measure |
| `flat_age` | `transaction_year` − `lease_commence_date` | Approximate flat age |
| `floor_level` | `storey_range` | Analyse floor-level effects |


### 5.1 Convert transaction date

```python
# Convert transaction month from string into datetime format
flat_data["transaction_date"] = pd.to_datetime(
    flat_data["month"], format="%Y-%m"
)

# Extract the transaction year
flat_data["transaction_year"] = flat_data["transaction_date"].dt.year

# Extract the calendar month number (1 to 12)
flat_data["month_number"] = flat_data["transaction_date"].dt.month

# Display the original and newly created date attributes
flat_data[
    ["month", "transaction_date", "transaction_year", "month_number"]
].head()
```

### 5.2 Convert remaining lease into numerical years

```python
# Extract the years and months from the remaining_lease string
lease_parts = flat_data["remaining_lease"].str.extract(
    r"(?P<years>\d+)\s+years?(?:\s+(?P<months>\d+)\s+months?)?"
)

# Convert the extracted year component into numeric values
lease_years = pd.to_numeric(lease_parts["years"], errors="coerce")

# Convert the extracted month component into numeric values
# Replace missing months with 0
lease_months = pd.to_numeric(
    lease_parts["months"].fillna("0"), errors="coerce"
)

# Convert the remaining lease into decimal years
flat_data["remaining_lease_years"] = (
    lease_years + lease_months / 12
)

# Display original and converted values for verification
flat_data[
    ["remaining_lease", "remaining_lease_years"]
].head(10)
```

```python
# Count any unsuccessful remaining lease conversions
print(
    "Invalid lease conversions:",
    flat_data["remaining_lease_years"].isna().sum()
)

# Check whether any month component falls outside 0 to 11
print(
    "Invalid lease month components:",
    (~lease_months.between(0, 11)).sum()
)
```

### 5.3 Calculate resale price per square metre

```python
# Calculate resale price per square metre
flat_data["price_per_sqm"] = (
    flat_data["resale_price"] / flat_data["floor_area_sqm"]
)

# Display the calculated price per square metre
flat_data[
    ["resale_price", "floor_area_sqm", "price_per_sqm"]
].head()
```

### 5.4 Calculate approximate flat age
Approximate age based on the lease commencement year, not necessarily the building's exact construction or occupation date.

```python
flat_data["flat_age"] = (
    flat_data["transaction_year"]
    - flat_data["lease_commence_date"]
)

flat_data[
    ["transaction_year", "lease_commence_date",
     "flat_age", "remaining_lease_years"]
].head()
```

### 5.5 Convert storey range into floor midpoint
`storey_range` contains values such as 01 TO 03, 04 TO 06, and 10 TO 12.
Since the exact floor is not given, we can calculate its midpoint as reference here.

`floor_level` is an estimated midpoint, not the exact floor on which a flat was sold.

```python
# Extract the min and max floor numbers
storey_parts = flat_data["storey_range"].str.extract(
    r"(\d+)\s+TO\s+(\d+)"
)

# Convert min and max floor to numeric
minimum_floor = pd.to_numeric(
    storey_parts[0], errors="coerce"
)
maximum_floor = pd.to_numeric(
    storey_parts[1], errors="coerce"
)

# Calculate midpoint of each storey range
flat_data["floor_level"] = (
    minimum_floor + maximum_floor
) / 2

flat_data[["storey_range", "floor_level"]].head()
```

# 7: Relevant Chart Creation

### 7.1 Monthly Resale Price Trends
How have HDB resale prices changed from 2017 to 2026, and do different flat types follow similar trends?

```python
selected_flat_types = ["3 ROOM", "4 ROOM", "5 ROOM"]

# Filter the dataset to include only these flat types
selected_data = flat_data[
    flat_data["flat_type"].isin(selected_flat_types)
].copy()

monthly_prices = (
    selected_data
    .groupby(["transaction_date", "flat_type"], as_index=False)
    ["resale_price"]
    .median()
)

plt.figure(figsize=(12, 6))

sb.lineplot(
    data=monthly_prices,
    x="transaction_date",
    y="resale_price",
    hue="flat_type",
    linewidth=2.5
)

plt.title("Monthly Median HDB Resale Prices by Flat Type")
plt.xlabel("Transaction Date")
plt.ylabel("Median Resale Price (SGD)")

plt.legend(title="Flat Type")
plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()
```

```python
yearly_price = (
    flat_data.groupby("transaction_year", as_index=False)["price_per_sqm"]
    .median()
)

plt.figure(figsize=(10, 5))

sb.lineplot(
    data=yearly_price,
    x="transaction_year",
    y="price_per_sqm",
    marker="o"
)

plt.title("Median HDB Resale Price per sqm Over Time")
plt.xlabel("Transaction Year")
plt.ylabel("Median Resale Price per sqm (SGD)")
plt.show()
```

```python
town_yearly_price = (
    flat_data
    .groupby(["transaction_year", "town"], as_index=False)["price_per_sqm"]
    .median()
)

# 2. Extract top 5 towns based on 2025 median price
top_5_towns = (
    town_yearly_price[town_yearly_price["transaction_year"] == 2025]
    .nlargest(5, "price_per_sqm")["town"]
    .tolist()
)

top5_yearly_price = town_yearly_price[town_yearly_price["town"].isin(top_5_towns)]

# 3. Create explicit color dictionary mapping each town to a distinct color
colors = sb.color_palette("tab10", n_colors=5)
palette_dict = dict(zip(top_5_towns, colors))

# 4. Split data for solid vs dashed segments
df_solid = top5_yearly_price[top5_yearly_price["transaction_year"] <= 2025]
df_dashed = top5_yearly_price[top5_yearly_price["transaction_year"] >= 2025]

plt.figure(figsize=(12, 6))

# Plot solid lines using explicit palette
ax = sb.lineplot(
    data=df_solid,
    x="transaction_year",
    y="price_per_sqm",
    hue="town",
    palette=palette_dict,
    marker="o",
    linestyle="-"
)

# Overlay dashed lines using the SAME palette dict
sb.lineplot(
    data=df_dashed,
    x="transaction_year",
    y="price_per_sqm",
    hue="town",
    palette=palette_dict,
    marker="o",
    linestyle="--",
    legend=False,
    ax=ax
)

plt.title("Top 5 Towns by Median Resale Price per sqm Over Time")
plt.xlabel("Transaction Year")
plt.ylabel("Median Resale Price per sqm (SGD)")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", title="Town")

plt.tight_layout()
plt.show()
```

### 7.2 Resale Price by HDB Town
How do HDB resale prices differ geographically?

```python
# Calculate the median resale price per sqm for each town
town_prices = (
    flat_data
    .groupby("town", as_index=False)["price_per_sqm"]
    .median()
)

# Sort towns from highest to lowest median price
town_prices = town_prices.sort_values(
    by="price_per_sqm",
    ascending=False
)

plt.figure(figsize=(12, 9))

sb.barplot(
    data=town_prices,
    x="price_per_sqm",
    y="town",
    color="steelblue"
)

plt.title("Median HDB Resale Price per sqm by Town")
plt.xlabel("Median Resale Price per sqm (SGD)")
plt.ylabel("HDB Town")

plt.tight_layout()
plt.show()
```

### 7.3 Remaining Lease vs Resale Price
Do flats with shorter remaining leases tend to sell for less?

graph below shows onnly year 2025, can we have an animation to loop from 2017 to 2025?

```python
# Select only 4-room flats 
lease_data = flat_data[
    flat_data["flat_type"] == "4 ROOM"
].copy()

# Limit the analysis to one complete calendar year
lease_data = lease_data[
    lease_data["transaction_year"] == 2025
]

# Randomly sample up to 5000 transactions
# This reduces overlapping observations in the scatterplot
lease_sample = lease_data.sample(
    n=min(5000, len(lease_data)),
    random_state=42
)

plt.figure(figsize=(11, 6))
sb.scatterplot(
    data=lease_sample,
    x="remaining_lease_years",
    y="price_per_sqm",
    alpha=0.35,
    s=25
)

plt.xlabel("Remaining Lease (Years)")
plt.ylabel("Resale Price per sqm (SGD)")

plt.title(
    "Remaining Lease vs Resale Price per sqm"
    "\n4-Room HDB Flats, 2025"
)

plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

### 7.4 Town × Flat Type Heatmap
Which combinations of HDB town and flat type have higher resale prices?

Are certain towns consistently expensive regardless of flat type? Are there exceptions?

```python
# Focus on one complete year
heatmap_data = flat_data[
    flat_data["transaction_year"] == 2025
].copy()

# Select common flat types
heatmap_data = heatmap_data[
    heatmap_data["flat_type"].isin(
        ["3 ROOM", "4 ROOM", "5 ROOM"]
    )
]

# Calculate the median price per sqm for each town and flat type
price_matrix = heatmap_data.pivot_table(
    index="town",
    columns="flat_type",
    values="price_per_sqm",
    aggfunc="median"
)

plt.figure(figsize=(10, 10))
sb.heatmap(
    price_matrix,
    cmap="Blues",
    annot=True,
    fmt=".0f",
    linewidths=0.4,
    cbar_kws={"label": "Median Price per sqm (SGD)"}
)

plt.title("Median Resale Price per sqm by Town and Flat Type (2025)")
plt.xlabel("Flat Type")
plt.ylabel("HDB Town")
plt.tight_layout()
plt.show()
```

### 7.4 Remaining Lease Across Different (Selected) Towns
Does the relationship between remaining lease and resale price differ across towns?

```python
# Select four towns for comparison
selected_towns = [
    "ANG MO KIO",
    "BEDOK",
    "QUEENSTOWN",
    "YISHUN"
]

# Filter by selected towns, flat type, and year
comparison_data = flat_data[
    (flat_data["town"].isin(selected_towns))
    & (flat_data["flat_type"] == "4 ROOM")
    & (flat_data["transaction_year"] == 2025)
].copy()

g = sb.lmplot(
    data=comparison_data,
    x="remaining_lease_years",
    y="price_per_sqm",
    col="town",
    col_wrap=2,
    height=4,
    aspect=1.2,
    scatter_kws={"alpha": 0.3},
    line_kws={"color": "red"}
)

g.set_axis_labels(
    "Remaining Lease (Years)",
    "Resale Price per sqm (SGD)"
)

g.set_titles("{col_name}")
g.fig.suptitle(
    "Remaining Lease vs Resale Price Across Selected Towns"
    "\n4-Room HDB Flats, 2025",
    y=1.04
)

plt.show()
```
